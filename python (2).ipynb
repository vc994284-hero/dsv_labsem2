```python
#list Method
# Initialize a base list
fruits = ["apple", "banana", "cherry"]

# 1. append() - Add to the end
fruits.append("orange")
print(fruits)  # ['apple', 'banana', 'cherry', 'orange']

# 2. extend() - Add multiple items
fruits.extend(["grape", "mango"])
print(fruits)  # ['apple', 'banana', 'cherry', 'orange', 'grape', 'mango']

# 3. insert() - Add at index 1
fruits.insert(1, "blueberry")
print(fruits)  # ['apple', 'blueberry', 'banana', 'cherry', 'orange', 'grape', 'mango']

# 4. remove() - Remove by value
fruits.remove("banana")
print(fruits)  # ['apple', 'blueberry', 'cherry', 'orange', 'grape', 'mango']

# 5. pop() - Remove and return the last item
last_fruit = fruits.pop()
print(last_fruit)  # 'mango'
print(fruits)      # ['apple', 'blueberry', 'cherry', 'orange', 'grape']

# 6. index() - Find position of a value
position = fruits.index("cherry")
print(position)  # 2

# 7. count() - Count occurrences
fruits.append("apple")  # Temporary add to show count
print(fruits.count("apple"))  # 2
fruits.pop()  # Clean up

# 8. sort() - Arrange alphabetically
fruits.sort()
print(fruits)  # ['apple', 'blueberry', 'cherry', 'grape', 'orange']

# 9. reverse() - Reverse the current order
fruits.reverse()
print(fruits)  # ['orange', 'grape', 'cherry', 'blueberry', 'apple']

# 10. copy() - Duplicate the list
fruits_backup = fruits.copy()
print(fruits_backup)  # ['orange', 'grape', 'cherry', 'blueberry', 'apple']

# 11. clear() - Wipe the list clean
fruits.clear()
print(fruits)  # []
```

```python
#dictionary
#Dictionaries are used to store data values in key:value pairs.
dict1 = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
print(dict1)
```

```python
# 1. Initialize and add items
user = {"name": "Alice", "age": 25}
user["city"] = "New York"

# 2. Safe access
print(user.get("status", "Not Found"))  # Output: Not Found

# 3. Merging dictionaries
extra_info = {"job": "Engineer", "age": 26}
user.update(extra_info)  

# 4. Iteration
for key, value in user.items():
    print(f"{key}: {value}")
```

```python
#printing patterns
#Right-Angled Triangle Pattern in Python
rows = int(input("Enter the row size for the pattern: "))
for i in range(1, rows + 1):  # Outer loop for rows
    for j in range(1, i + 1):  # Inner loop for columns
        print("*", end=" ")   # Print star
    print()
```
