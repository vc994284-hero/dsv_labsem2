```python
import numpy as np
import pandas as pd

travel_df = pd.DataFrame({
    'Place':['Delhi','Mumbai','Pune','Chennai','Kolkata'],
    'Days':[2,3,4,5,6]
})

travel_df['Code'] = np.arange(5)

print("After np.arange():")
print(travel_df)

code_series = pd.Series(
    ['A101','B202'],
    index=[1,4]
)

travel_df['Code'] = code_series

print("\nAfter Series Assignment:")
print(travel_df)
```

```python
school_data = {
    'Student':['A','B','C','D','E'],
    'Class':[10,10,9,8,7],
    'Section':['A','B','A','C','B']
}

school_df = pd.DataFrame(
    school_data,
    columns=['Student','Class','Section','House']
)

print(school_df)

school_df['House'] = 'Red'

print("\nAfter assigning House:")
print(school_df)
```

```python
inventory_df = pd.DataFrame({
    'Item':['Pen','Book','Laptop','Mouse','Keyboard'],
    'Quantity':[50,20,10,15,8]
})

supplier = pd.Series(
    ['ABC Ltd','XYZ Ltd'],
    index=[1,3]
)

inventory_df['Supplier'] = supplier

print("After adding Supplier:")
print(inventory_df)

del inventory_df['Supplier']

print("\nAfter deleting Supplier:")
print(inventory_df)
```

```python
import io
import pandas as pd

# Mock representation of the copied data
clipboard_data = """Name,Total Marks,Age
Amit,85,20
Riya,91,21
Karan,78,19
Neha,89,22
Vikas,95,24
"""

# Read copied table
lab_df = pd.read_csv(io.StringIO(clipboard_data))

# Display DataFrame
print(lab_df)

# Display column names
print(lab_df.columns)

# Retrieve one single-word column
print(lab_df['Name'])

# Retrieve one column with space in its name
print(lab_df['Total Marks'])

# Smaller DataFrame
small_df = lab_df[['Name','Total Marks','Age']]
print(small_df)

# Add new column
lab_df['Status'] = 'Active'
print(lab_df)

# Delete column
del lab_df['Status']
print(lab_df)
```
