```python
#que 19:
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

sales_2025 = [100, 150, 120, 180, 200]
sales_2026 = [150, 180, 170, 220, 280]

growth = [
    sales_2026[i] - sales_2025[i]
    for i in range(len(months))
]

max_index = growth.index(max(growth))

colors = ['red' if i == max_index else 'blue'
          for i in range(len(growth))]

bars = plt.bar(months, growth, color=colors)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        str(bar.get_height()),
        ha='center'
    )

plt.title("Monthly Sales Growth")
plt.xlabel("Months")
plt.ylabel("Sales Growth")

plt.show()
```

```python
#que 20:
import matplotlib.pyplot as plt
import statistics

marks = [45, 55, 60, 65, 70, 75, 80, 85, 90, 95]

mean = statistics.mean(marks)
median = statistics.median(marks)

plt.hist(marks, bins=8, edgecolor='black')

plt.axvline(mean, color='red', linestyle='--',
            label='Mean')

plt.axvline(median, color='green', linestyle=':',
            label='Median')

plt.title("Students Marks Report")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.legend()

print("Mean:", mean)
print("Median:", median)

plt.savefig("marks_report.png", dpi=200)

plt.show()
```
